# Tensile Testing Tutorial

Welcome to the tensile testing tutorial. This guide walks you through the three main phases of a tensile test:

1. **Pre-Test Setup**
2. **Test Setup**
3. **Post-Test Procedure**

## 1.  Pre-Test Setup

Before initiating a test, ensure all dimensions and markings are completed on the specimen.

### Steps:

- Measure and record:
  - Gauge width
  - Grip width
  - Grip length
  - Specimen thickness

- Mark the **gauge length** using a white pen:
  - Ensure that the gauge markers is within the length of the gauge region on the specimen

- Measure the grip handle on **one side**, then mirror that measurement and mark the **other side** of the specimen.

## 2.  Test Setup

Properly install the specimen and configure the test machine.

###  Mounting the Specimen:

- Insert the specimen into a jig, then load it into the **bottom crosshead**.
- Bring the **top clamp** down in an **open orientation** to avoid damaging the sample.
- Ensure the clamp aligns with the grip marks.
- Raise the crosshead slightly, then remove the jig.
- Lower the top clamp so that it grips the specimen at the correct mark.

### Displacement Rate:

- Ensure that the displacement rate is between 0.001 * gauge length - 0.01 * gauge length
    - This gives a discplacement rate with units mm/sec

###  Extensometer Setup:

- Position the extensometer to align exactly with the **white gauge marks**.
- **Calibrate** the extensometer using the MTS system:
  - This allows the system to convert extensometer displacement into strain values accurately.

## 3. Post-Test Procedure

Once the specimen has failed, follow these cleanup and recording steps:

###  Post-Test Checks:

- Confirm that failure occurred **within the gauge region** of the specimen.
- Carefully remove the specimen from both the top and bottom clamps.
  - Avoid additional damage to the fractured surfaces.

###  Handling and Labeling:

- Wrap the tested specimen in **masking tape**.
- Label the sample properly with the correct **ID or description**.

##  Notes

- Do not reuse broken or deformed grips — this can affect the test validity.
- If the specimen breaks **outside the gauge region**, the test result may be considered invalid.
- Always document any anomalies during the test (e.g., slipping, premature failure, or cracking sounds near the grip).
